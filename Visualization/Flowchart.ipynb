### 1. English Flowchart (Ultra-Compact For A4 Paper)
To absolutely guarantee it fits on an A4 page without taking up much space, we applied the standard scientific practice of **Loop Condensation**. 
Instead of drawing 5 separate condition nodes for 5 nested loops, we group the **Data Loops (k, N, r)** and the **Model Loops (G, M)**. 
This reduces the diagram from ~20 nodes down to **just 9 nodes**, making it extremely compact while preserving 100% of the algorithmic logic.

```mermaid
flowchart TD
    Start([Start]) --> Init[/"Input 𝓓, 𝓝, 𝓢, 𝓖, &Phi;<sub>G</sub>, 𝓜"/]
    Init --> Step1["Partition 𝓓 into 10 folds <i>D<sup>k</sup></i>"]
    
    Step1 --> L1{"<b>Outer Loops</b><br>For k = 1..10<br>For <i>N</i> &isin; 𝓝<br>For r = 1..10"}
    
    L1 -- "Next Data Split" --> Prep["Set <i>D<sup>k</sup><sub>test</sub></i> and <i>D<sup>k</sup><sub>pool</sub></i><br>Sample <i>N</i> per class from <i>D<sup>k</sup><sub>pool</sub></i> &rarr; <i>D<sup>(r)</sup><sub>seed</sub></i>"]
    
    Prep --> L2{"<b>Inner Loops</b><br>For Configuration (<i>G, &phi;, S</i>)<br>For Classifier <i>M</i> &isin; 𝓜"}
    
    L2 -- "Next Model" --> Train["1. Train <i>G(&phi;)</i> with <i>D<sup>(r)</sup><sub>seed</sub></i> &rarr; Generate <i>D<sub>synth</sub></i><br>2. Hyperparameter Random Search (Validate on <i>D<sup>(r)</sup><sub>seed</sub></i>) &rarr; <i>&theta;<sub>opt</sub></i><br>3. Train <i>M</i> using <i>&theta;<sub>opt</sub></i>, Evaluate on <i>D<sup>k</sup><sub>test</sub></i>"]
    
    %% Loops
    Train --> L2
    L2 -- "End Models" --> L1
    
    L1 -- "End Data Loops" --> Step5["Aggregate scores &rarr; Matrix <i>P</i>"]
    Step5 --> Output[/"Output <i>P</i>"/]
    Output --> End([End])

    classDef loop fill:#f9f9f9,stroke:#555,stroke-width:2px;
    class L1,L2 loop;
```

````python
import base64
import urllib.request
import webbrowser
import os
from IPython.display import Image, display

graph = """
flowchart TD
    Start([Start]) --> Init[/"Input 𝓓, 𝓝, 𝓢, 𝓖, &Phi;<sub>G</sub>, 𝓜"/]
    Init --> Step1["Partition 𝓓 into 10 folds <i>D<sup>k</sup></i>"]
    
    Step1 --> L1{"<b>Outer Loops</b><br>For k = 1..10<br>For <i>N</i> &isin; 𝓝<br>For r = 1..10"}
    
    L1 -- "Next Data Split" --> Prep["Set <i>D<sup>k</sup><sub>test</sub></i> and <i>D<sup>k</sup><sub>pool</sub></i><br>Sample <i>N</i> per class from <i>D<sup>k</sup><sub>pool</sub></i> &rarr; <i>D<sup>(r)</sup><sub>seed</sub></i>"]
    
    Prep --> L2{"<b>Inner Loops</b><br>For Configuration (<i>G, &phi;, S</i>)<br>For Classifier <i>M</i> &isin; 𝓜"}
    
    L2 -- "Next Model" --> Train["1. Train <i>G(&phi;)</i> with <i>D<sup>(r)</sup><sub>seed</sub></i> &rarr; Generate <i>D<sub>synth</sub></i><br>2. Hyperparameter Random Search (Validate on <i>D<sup>(r)</sup><sub>seed</sub></i>) &rarr; <i>&theta;<sub>opt</sub></i><br>3. Train <i>M</i> using <i>&theta;<sub>opt</sub></i>, Evaluate on <i>D<sup>k</sup><sub>test</sub></i>"]
    
    Train --> L2
    L2 -- "End Models" --> L1
    
    L1 -- "End Data Loops" --> Step5["Aggregate scores &rarr; Matrix <i>P</i>"]
    Step5 --> Output[/"Output <i>P</i>"/]
    Output --> End([End])
"""

b64 = base64.urlsafe_b64encode(graph.encode('utf8')).decode('ascii')
url = "https://mermaid.ink/img/" + b64
req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
image_filename = 'experiment_flowchart_latex.png'

try:
    with urllib.request.urlopen(req) as response, open(image_filename, 'wb') as f:
        f.write(response.read())
    display(Image(image_filename))
    webbrowser.open('file://' + os.path.realpath(image_filename))
except Exception as e:
    pass
````
