# Multi-process Orchestrator for Workstation

This notebook uses concurrent.futures to orchestrate automated experiment execution.
Instead of running sequentially or opening 10 notebooks manually which overloads the OS, this notebook acts as a Queue, limiting concurrent workers to maximize CPU resources without bottlenecks.

```python
import os
import subprocess
import concurrent.futures
from tqdm.notebook import tqdm
import sys

# Configure paths
EXPERIMENT_SCRIPT = os.path.join('..', 'experiment.py')
LOG_DIR = 'logs'
os.makedirs(LOG_DIR, exist_ok=True)

# Max parallel datasets
# NOTE: With i7 CPU 12700, config N_JOBS=3, worker: 20 // 3 = 6.
# If running on a large workstation (64 threads) and N_JOBS=-1, adjust MAX_WORKERS = threads // 10.
MAX_WORKERS = 10

DATASETS = ["Coffee", "Plant_oil", "Brewed_vinegar", "Wine_spoilage", "Chinese_wine"]
SEEDS = [1, 2, 5, 10, 20]
N_SPLITS = 5

def run_task(task_kwargs):
    dataset = task_kwargs['dataset']
    seed = task_kwargs['seed']
    fold = task_kwargs['fold']
    pbar = task_kwargs['pbar']
    log_file = os.path.join(LOG_DIR, f"{dataset}_S{seed}_F{fold}.log")
    
    with open(log_file, "w", encoding="utf-8") as f:
        f.write(f"Starting {dataset} - Seed {seed} - Fold {fold}\n\n")
        f.flush()
        
        command = [sys.executable, EXPERIMENT_SCRIPT, "--dataset", dataset, "--seed", str(seed), "--fold", str(fold), "--n_selections", "5"]
        process = subprocess.Popen(command, stdout=f, stderr=subprocess.STDOUT, text=True)
        process.wait()
        
        pbar.update(1)
        
        if process.returncode != 0:
            return f"[!] ERROR in {dataset} (Seed {seed}, Fold {fold}) - See details at {log_file}"
                
    return f"[+] COMPLETED {dataset} (Seed {seed}, Fold {fold})"
```

```python
# Create task list and shared progress bar
total_folds = len(DATASETS) * len(SEEDS) * N_SPLITS
pbar = tqdm(total=total_folds, desc="Total Folds Progress")

tasks = []
for dataset in DATASETS:
    for seed in SEEDS:
        for fold in range(1, N_SPLITS + 1):
            tasks.append({'dataset': dataset, 'seed': seed, 'fold': fold, 'pbar': pbar})
print(f"Total sequence tasks: {len(tasks)}")
print(f"Total individual folds to run: {total_folds}")
print(f"Parallel workers: {MAX_WORKERS}")
print("Starting... Output will be hidden and written to Runners/logs/")

results = []
# Use ThreadPoolExecutor for parallel execution to avoid ProcessPool bug on Windows Jupyter
with concurrent.futures.ThreadPoolExecutor(max_workers=MAX_WORKERS) as executor:
    futures = [executor.submit(run_task, task) for task in tasks]
    for future in concurrent.futures.as_completed(futures):
        try:
            res = future.result()
            print(res)
            results.append(res)
        except Exception as e:
            print(f"[!!!] Process exception: {e}")

pbar.close()
print("\n=== ALL TASKS COMPLETED ===")
```
